```python
import math
import subprocess
import numpy as np
import meep as mp
import re
from meep import mpb
from IPython.utils.capture import capture_output
from tqdm import tqdm
import time
import matplotlib.pyplot as plt
```

# Photonics

```python
def atlas_map(radius,epsilon):

    sqrt_half = math.sqrt(0.5)
    geometry_lattice = mp.Lattice(
        basis_size=mp.Vector3(sqrt_half, sqrt_half, sqrt_half),
        basis1=mp.Vector3(0, 1, 1),
        basis2=mp.Vector3(1, 0, 1),
        basis3=mp.Vector3(1, 1)
    )

    # Corners of the irreducible Brillouin zone for the fcc lattice,
    # in a canonical order:
    vlist = [
        mp.Vector3(0, 0.5, 0.5),        # X
        mp.Vector3(0, 0.625, 0.375),    # U
        mp.Vector3(0, 0.5, 0),          # L
        mp.Vector3(0, 0, 0),            # Gamma
        mp.Vector3(0, 0.5, 0.5),        # X
        mp.Vector3(0.25, 0.75, 0.5),    # W
        mp.Vector3(0.375, 0.75, 0.375)  # K
    ]
    # k_points = Bloch wavevectors
    k_points = mp.interpolate(4, vlist)

#    # define a couple of parameters (which we can set from the command_line)
#    eps = 11.56  # the dielectric constant of the spheres
#    r = 0.25  # the radius of the spheres

    diel = mp.Medium(epsilon=epsilon)

    # A diamond lattice has two "atoms" per unit cell:
    geometry = [mp.Sphere(radius, center=mp.Vector3(0.125, 0.125, 0.125), material=diel),
                mp.Sphere(radius, center=mp.Vector3(-0.125, -0.125, -0.125), material=diel)]

    # A simple fcc lattice would have only one sphere/object at the origin.

    resolution = 16  # use a 16x16x16 grid
    mesh_size = 5
    num_bands = 5
    # Define and run simulation for band structure calculation
    ms = mpb.ModeSolver(
        geometry_lattice=geometry_lattice,
        k_points=k_points,
        geometry=geometry,
        resolution=resolution,
        num_bands=num_bands,
        mesh_size=mesh_size
    )
    with capture_output() as output:
        #ms.run(mpb.output_at_kpoint(mp.Vector3(0, 0.625, 0.375), mpb.output_dpwr))
        ms.run_tm()
    variable_containing_output = output.stdout.strip()
    pattern = r'band\s+\d+\s*\(\s*([\d.]+)\s*\)'
    matches = re.findall(pattern, variable_containing_output)
    matches=[eval(i) for i in matches]
    return matches
```

```python
def extract(lst,n):
    return list(list(zip(*lst))[n])
def plot_atlas(epsilon):
    radius = np.arange(0,0.5,0.01)
    freq=[]
    for r in tqdm(radius):
        freq.append(atlas_map(r,epsilon))  
    for i in range(0,len(freq)):
        if(freq[i]==[]):
            freq[i]=[0,0]
    plt.xlabel('Radius')
    plt.ylabel('Frequencies')
    plt.title("Epsilon="+str(epsilon))

    plt.scatter(radius, extract(freq,0), marker='.', color="blue")
    plt.scatter(radius, extract(freq,1), marker='.', color="red")
    plt.show()
```

```python
for eps in np.arange(3,5,0.01):
    plot_atlas(eps)
```
