```python
for i in range(len(data.index)):
    data_anal = data.iloc[i]
    row_list = []

    df = pd.DataFrame(pd.unique(data['Space Group']), columns = ['Space Group'])
    space = [data_anal['Space Group']]
    row_list.extend(df['Space Group'].isin(space).astype(int).tolist())

    df = pd.DataFrame(pd.unique(data['Global Point Group']), columns = ['Global Point Group'])
    glob = [data_anal['Global Point Group']]
    row_list.extend(df['Global Point Group'].isin(glob).astype(int).tolist())

    global_ = data_anal['Global Point Group']
    df = pd.DataFrame(pd.unique(data['Global Point Group']), columns = ['Inherited Global Point Group'])
    row_list.extend(df['Inherited Global Point Group'].isin(cell[global_]).astype(int).tolist())
    
    df = pd.DataFrame(pd.unique(data['Global Point Group']), columns = ['Local Point Group'])
    local = data_anal['Local Point Groups'].replace(' ','').split(',')
    row_list.extend(df['Local Point Group'].isin(local).astype(int).tolist())
    
    local = data_anal['Local Point Groups'].replace(' ','').split(',')
    df = pd.DataFrame(pd.unique(data['Global Point Group']), columns = ['Inherited Local Point Group'])
    if len(local) == 1:
        row_list.extend(df['Inherited Local Point Group'].isin(local).astype(int).tolist())
    else:
        inhet_local = []
        for k in range(len(local)):
            if k == 0:
                inhet_local.extend(df['Inherited Local Point Group'].isin(cell[local[k]]).astype(int).tolist())
            elif k > 0:
                inhet_local2 = df['Inherited Local Point Group'].isin(cell[local[k]]).astype(int).tolist()
                for j in range(len(inhet_local2)):
                    if inhet_local[j] != inhet_local2[j] and inhet_local2[j] == 1:
                        inhet_local[j] = inhet_local2[j]
        row_list.extend(inhet_local)

    df = pd.DataFrame(pd.unique(data['Lattice Type']), columns = ['Lattice Type'])
    lattice = [data_anal['Lattice Type']]
    row_list.extend(df['Lattice Type'].isin(lattice).astype(int).tolist())

    df = pd.DataFrame(pd.unique(data['Centering']), columns = ['Centering'])
    cent = [data_anal['Centering']]
    row_list.extend(df['Centering'].isin(cent).astype(int).tolist())

    df_new.loc[i] = row_list

df_new.to_csv('crystal_info_including_inherited.csv', index=False)
```
